TASK 2 ASSIGNMENT 20

```python
for col in text_columns:
    df[col] = df[col].apply(lambda x:x.lower() if isinstance(x, str) else x)
```

```python
df[text_columns].head()
```

```python
import re
```

```python
for col in text_columns:
    df[col] = df[col].apply(lambda x: re.sub(r'[^a-zA-Z0-9\s]', '', x)if isinstance(x, str) else x)
```

```python
df[text_columns].head()
```

```python
import nltk 
from nltk.corpus import stopwords
nltk.download('stopwords')
stop_words = set(stopwords.words('english'))
```

```python
for col in text_columns:
    df[col] = df[col].apply(lambda x: ' '.join(word for word in x.split() if word not in stop_words) if isinstance(x, str)else x)
```

```python
df[text_columns].head()
```

```python
for col in text_columns:
    df[col] = df[col].fillna('')
```

```python
df[text_columns].isnull().sum()
```

```python
df['clean_text'] = df[text_columns].astype(str).agg(' '.join, axis = 1)
```

```python
df['clean_text'].head()
```

TASK 3 ASSIGNMENT 20

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf = TfidfVectorizer(max_features = 5000, ngram_range = (1, 2))
```

```python
tfidf_matrix = tfidf.fit_transform(df['clean_text'])
```

```python
tfidf_matrix.shape
```

TASK 4 ASSIGNMENT 20

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity_matrix = cosine_similarity(tfidf_matrix)
```

```python
similarity_matrix.shape
```

```python
import numpy as np
```

```python
np.save('similarity_matrix.npy', similarity_matrix)
print("Similarity matrix saved successfully!")
```

```python
similarity_matrix[:5, :5]
```

Why cosine similarity is used?
Cosine similarity measures how similar two movies are based on their text features. It compares the angle between their TF-IDF vectors, helping identify movies with similar descriptions, genres, and keywords. A higher score indicates greater similarity, making it useful for content-based movie recommendation systems.

```python
import streamlit as st
```

```python
similarity_matirx = np.load("similarity_matrix.npy")
```

```python
def recommend(item_name, top_n = 5):
    if item_name not in df['title'].values:
        return[]
    index = df[df['title']==item_name].index[0]
    scores = list(enumerate(similarity_matirx[index]))
    scores= sorted(scores, key=lambda x: x[1], reverse=True)
    scores = [s for s in scores if s[0] != index]
    return [(df.iloc[i]['title'], round(score, 3)) for i, score in scores[:top_n]]
```

```python
st.title("movie recommendation system")
movie_name = st.selectbox("Select a movie", df['title'].dropna().unique())
if st.button("Recommend Movies"):
    recommendations = recommend(movie_name)
    st.subheader("recommend Movies")
    if recommendations:
        for movie, score in recommendations:
            st.write(f"{movie}-Similarity:{score}")
        else:
            st.warning("No recommendations found")
```
